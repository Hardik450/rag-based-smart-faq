```python
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
import os

load_dotenv()
google_api_key = os.getenv('GOOGLE_API_KEY')
chat = ChatGoogleGenerativeAI(model = 'gemini-2.0-flash', temperature = 0.7)
```

```python
from langchain.document_loaders import PyMuPDFLoader

file = 'D:\cs50\gfg tensorflow\Syllabus 3rd Year AIDS V & VI Sem..pdf'
loader = PyMuPDFLoader(file_path = file)
docx = loader.load()

print(docx[0].page_content)
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain.schema import Document
import time

# ✅ Load HuggingFace Embeddings
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

from langchain.vectorstores import FAISS
db = FAISS.load_local("my_vector_db", embeddings, allow_dangerous_deserialization=True)
```

```python
retriever = db.as_retriever()

from langchain.chains import RetrievalQA

qa = RetrievalQA.from_chain_type(
    llm = chat, 
    retriever = retriever, 
    return_source_documents = True, 
    chain_type = "stuff", 
    verbose = True
    )
```

```python
q = "How many units are there in Computer Graphics and Multimedia? Name them all."
response = qa.invoke({"query":q})
print(response)
```
